```python
"""Customer churn prediction: continues from your EDA script.

Run:  python churn_model.py
Needs: pandas numpy scikit-learn matplotlib joblib
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, RocCurveDisplay, precision_recall_curve,
)

RANDOM_STATE = 42

# ---------------------------------------------------------------- 1. Load + clean
data = pd.read_csv("telecom_churn_data.csv")

# TotalCharges has blank strings for brand-new customers -> convert to numeric
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"], errors="coerce")

# customerID is an identifier, not a feature
data = data.drop(columns=[c for c in ["customerID"] if c in data.columns])

# Target: Yes/No -> 1/0
y = (data["Churn"] == "Yes").astype(int)
X = data.drop(columns="Churn")

# ---------------------------------------------------------------- 2. Split first (avoids leakage)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

# ---------------------------------------------------------------- 3. Preprocessing
num_cols = X.select_dtypes(include="number").columns.tolist()
cat_cols = X.select_dtypes(exclude="number").columns.tolist()

# SeniorCitizen is 0/1 already; it sits in num_cols, which is fine.
from sklearn.impute import SimpleImputer

preprocess = ColumnTransformer([
    ("num", Pipeline([
        ("impute", SimpleImputer(strategy="median")),
        ("scale", StandardScaler()),
    ]), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

# ---------------------------------------------------------------- 4. Candidate models
# class_weight="balanced" compensates for churners being the minority (~26%)
models = {
    "LogisticRegression": LogisticRegression(max_iter=1000, class_weight="balanced"),
    "RandomForest": RandomForestClassifier(
        n_estimators=300, min_samples_leaf=3, class_weight="balanced",
        n_jobs=-1, random_state=RANDOM_STATE),
    "HistGradientBoosting": HistGradientBoostingClassifier(
        learning_rate=0.05, max_iter=300, class_weight="balanced",
        random_state=RANDOM_STATE),
}

pipes = {name: Pipeline([("prep", preprocess), ("model", m)]) for name, m in models.items()}

# ---------------------------------------------------------------- 5. Compare with cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
print("5-fold CV ROC-AUC (train set only)")
scores = {}
for name, pipe in pipes.items():
    s = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="roc_auc")
    scores[name] = s.mean()
    print(f"  {name:22s} {s.mean():.3f} +/- {s.std():.3f}")

best_name = max(scores, key=scores.get)
best = pipes[best_name]
print(f"\nBest model: {best_name}")

# ---------------------------------------------------------------- 6. Final fit + test evaluation
best.fit(X_train, y_train)
proba = best.predict_proba(X_test)[:, 1]
pred = (proba >= 0.5).astype(int)

print(f"\nTest ROC-AUC: {roc_auc_score(y_test, proba):.3f}")
print(classification_report(y_test, pred, target_names=["Stay", "Churn"]))

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ConfusionMatrixDisplay(confusion_matrix(y_test, pred), display_labels=["Stay", "Churn"]).plot(ax=ax[0])
RocCurveDisplay.from_predictions(y_test, proba, ax=ax[1])
plt.tight_layout()
plt.savefig("evaluation.png", dpi=150)

# ---------------------------------------------------------------- 7. Pick a threshold that fits the business goal
# Example: we want to catch at least 80% of churners (recall), then see the precision cost.
prec, rec, thr = precision_recall_curve(y_test, proba)
ok = np.where(rec[:-1] >= 0.80)[0]
if len(ok):
    i = ok[-1]
    print(f"Threshold {thr[i]:.2f} -> recall {rec[i]:.2f}, precision {prec[i]:.2f}")

# ---------------------------------------------------------------- 8. What drives churn?
from sklearn.inspection import permutation_importance

r = permutation_importance(best, X_test, y_test, scoring="roc_auc",
                           n_repeats=5, random_state=RANDOM_STATE, n_jobs=-1)
imp = pd.Series(r.importances_mean, index=X_test.columns).sort_values(ascending=False)
print("\nTop drivers of churn (permutation importance):")
print(imp.head(8).round(4))

# ---------------------------------------------------------------- 9. Save and reuse
joblib.dump(best, "churn_model.joblib")

# Score new customers:
#   model = joblib.load("churn_model.joblib")
#   model.predict_proba(new_df)[:, 1]   # churn probability per customer
```
